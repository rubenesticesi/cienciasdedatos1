# INFORME DE ANALISIS EXPLORATORIO: Población Estudiantil con TEA en Cali

**INTRODUCCION**

Este informe presenta un análisis exploratorio de datos (EDA) sobre la población estudiantil diagnosticada con Trastorno del Espectro Autista (TEA) en la ciudad de Cali, Colombia, para el año 2025. El objetivo es normalizar la base de datos, analizar las características demográficas y educativas de esta población, identificar patrones y posibles sesgos, y derivar conclusiones que puedan apoyar la toma de decisiones en políticas públicas y la asignación de recursos.

**SELECCION DE VAARIABLES CLAVE Y SU IMPORTANCIA**

Para este análisis descriptivo, no se busca predecir una única variable, sino caracterizar a la población. Por lo tanto, nos enfocaremos en un conjunto de variables clave:

**Edad**: Fundamental para entender en qué etapas del ciclo de vida y del sistema educativo se concentra la población.

**Género**: Clave para analizar la distribución y prevalencia del TEA entre hombres y mujeres, un factor ampliamente estudiado.

**Comuna**: Esencial para el análisis geoespacial, permitiendo identificar la distribución de la población en la ciudad y focalizar recursos.

**Sector y Zona Educativa**: Variables críticas para comprender si los estudiantes están siendo atendidos mayoritariamente en el sistema público o privado, y en zonas urbanas o rurales.

Estas variables, en conjunto, nos darán una radiografía completa de la población estudiantil con TEA en Cali.

#1. - Código Python: Análisis Exploratorio de Datos - Población TEA en Cali 2025   
#2. - Cargar Archivo, Librerias y Normalización

**Para el dataset POBLACIÖN TEA CALI.xlsx**

**La variable numérica más importante es EDAD**

En un análisis descriptivo cuyo objetivo es caracterizar a una población, la EDAD es fundamental. Permite entender la estructura demográfica del grupo, identificar en qué etapas del ciclo educativo se concentran los estudiantes (en este caso, la primaria) y planificar los recursos y apoyos necesarios para cada grupo etario. A diferencia de los códigos o identificadores, la EDAD ofrece un profundo contexto sobre la población estudiantil.

```python
"""
Análisis Exploratorio de Datos - Población TEA en Cali 2025
"""

# --- PASO 1: Importar librerías necesarias ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import statsmodels.api as sm

# Configurar el estilo de los gráficos
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['axes.titlesize'] = 16
plt.rcParams['axes.labelsize'] = 12


# --- PASO 2: Carga y Normalización de la Base de Datos ---
print("--- Iniciando Normalización de Datos ---")

# Cargar el archivo CSV
try:
    df = pd.read_excel('POBLACIÖN TEA CALI.xlsx')
    print("Dataset cargado exitosamente.")
except FileNotFoundError:
    print("Error: No se encontró el archivo 'POBLACIÖN TEA CALI.xlsx'.")
    exit()

# a) Selección de columnas relevantes y eliminación de PII (Información Personalmente Identificable)
columnas_relevantes = [
    'COMUNA', 'SECTOR', 'ZONA', 'FECHA_NACIMIENTO', 'GENERO',
    'ETNIA', 'TIPO_DISCAPACIDAD', 'GRADO', 'CARACTER', 'ESPECIALIDAD'
]
df = df[columnas_relevantes]

# b) Cálculo de la Edad (Feature Engineering)
# Convertir 'FECHA_NACIMIENTO' a formato fecha, manejando errores
df['FECHA_NACIMIENTO'] = pd.to_datetime(df['FECHA_NACIMIENTO'], errors='coerce')
# Eliminar filas donde la fecha no pudo ser convertida
df.dropna(subset=['FECHA_NACIMIENTO'], inplace=True)
# Calcular la edad basada en el año actual (2025)
current_year = 2025
df['EDAD'] = current_year - df['FECHA_NACIMIENTO'].dt.year

# c) Estandarización de Variables Categóricas
df['GENERO'] = df['GENERO'].replace({'F': 'Femenino', 'M': 'Masculino'})
df['SECTOR'] = df['SECTOR'].replace({'NO OFICIAL': 'No Oficial', 'OFICIAL': 'Oficial'})
df['ZONA'] = df['ZONA'].replace({'URBANO': 'Urbana', 'RURAL': 'Rural'})

# d) Limpieza de datos faltantes y valores inconsistentes
# Reemplazar valores nulos en 'COMUNA' con 'No especificada' para el análisis
df['COMUNA'].fillna('No especificada', inplace=True)
# Asegurar que 'COMUNA' sea de tipo string para graficar
df['COMUNA'] = df['COMUNA'].astype(str)

print("Forma del dataset limpio:", df.shape)
print("\n--- Normalización Completada ---\n")





```

```python
# --- PASO 3: Análisis Univariado ---
print("--- Realizando Análisis Univariado ---")

# a) Distribución de Edad
plt.figure()
sns.histplot(df['EDAD'], kde=True, bins=20, color='skyblue')
plt.title('Distribución de Edades de la Población con TEA')
plt.xlabel('Edad (años)')
plt.ylabel('Frecuencia')
plt.show()

# La variable numérica más importante es EDAD.

# Variable a analizar
df['EDAD']
# Calcular estadísticas descriptivas

media = df['EDAD'].mean()
mediana = df['EDAD'].median()
moda = df['EDAD'].mode().iloc[0] if not df['EDAD'].mode().empty else None
std_dev = df['EDAD'].std()

# Detectar valores atípicos usando el método de IQR
Q1 = df['EDAD'].quantile(0.25)
Q3 = df['EDAD'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = df[(df['EDAD'] < lower_bound) | (df['EDAD'] > upper_bound)]['EDAD']


# Resultados
resultados = {
   'media': media,
   'mediana': mediana,
   'moda': moda,
   'desviacion_estandar': std_dev,
   'limite_inferior_outliers': lower_bound,
   'limite_superior_outliers': upper_bound,
   'valores_atipicos': outliers.tolist()
        }

print(resultados)


# b) Distribución de Género
plt.figure()
sns.countplot(x='GENERO', data=df, palette='viridis', order=df['GENERO'].value_counts().index)
plt.title('Distribución por Género')
plt.xlabel('Género')
plt.ylabel('Cantidad de Estudiantes')
plt.show()

# c) Distribución por Comuna
plt.figure(figsize=(14, 8))
sns.countplot(y='COMUNA', data=df, palette='plasma', order=df['COMUNA'].value_counts().index)
plt.title('Distribución de Estudiantes por Comuna')
plt.xlabel('Cantidad de Estudiantes')
plt.ylabel('Comuna')
plt.show()

# d) Distribución por Sector Educativo
plt.figure()
sns.countplot(x='SECTOR', data=df, palette='magma', order=df['SECTOR'].value_counts().index)
plt.title('Distribución por Sector Educativo (Oficial vs. No Oficial)')
plt.xlabel('Sector')
plt.ylabel('Cantidad de Estudiantes')
plt.show()
```

# matriz de correlacion
Código Python para el Análisis de Correlación
Este script primero limpia y prepara los datos (un paso crucial para la correlación) y luego calcula y visualiza la matriz de correlación en un mapa de calor.

```python
# --- PASO 1: Importar librerías necesarias ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- PASO 2: Cargar y Preparar los Datos ---
print("--- Cargando y preparando los datos... ---")
try:
    df = pd.read_csv('POBLACIÖN TEA CALI.xlsx - Hoja1.csv')
except FileNotFoundError:
    print("Error: No se encontró el archivo del dataset.")
    exit()

# Seleccionar columnas relevantes y eliminar PII
columnas_relevantes = [
    'COMUNA', 'SECTOR', 'ZONA', 'FECHA_NACIMIENTO', 'GENERO',
    'ETNIA', 'GRADO', 'CARACTER'
]
df_clean = df[columnas_relevantes].copy() # Usar .copy() para evitar advertencias

# Calcular la Edad
df_clean['FECHA_NACIMIENTO'] = pd.to_datetime(df_clean['FECHA_NACIMIENTO'], errors='coerce')
df_clean.dropna(subset=['FECHA_NACIMIENTO'], inplace=True)
df_clean['EDAD'] = 2025 - df_clean['FECHA_NACIMIENTO'].dt.year
df_clean.drop('FECHA_NACIMIENTO', axis=1, inplace=True) # Eliminar la columna de fecha original

# Estandarizar y limpiar datos categóricos
df_clean['GENERO'] = df_clean['GENERO'].replace({'F': 'Femenino', 'M': 'Masculino'})
df_clean['SECTOR'] = df_clean['SECTOR'].replace({'NO OFICIAL': 'No Oficial', 'OFICIAL': 'Oficial'})
df_clean['ZONA'] = df_clean['ZONA'].replace({'URBANO': 'Urbana', 'RURAL': 'Rural'})
df_clean.dropna(inplace=True) # Eliminar filas con datos faltantes para un análisis limpio

print("--- Datos preparados. Procediendo con el análisis de correlación... ---")

# --- PASO 3: Convertir Variables Categóricas a Numéricas ---
# La correlación solo funciona con números. Convertimos cada categoría a un código numérico.
df_corr = df_clean.copy()
for column in df_corr.select_dtypes(include=['object']).columns:
    df_corr[column] = df_corr[column].astype('category').cat.codes

# --- PASO 4: Calcular la Matriz de Correlación ---
correlation_matrix = df_corr.corr()

# --- PASO 5: Visualizar la Matriz con un Mapa de Calor (Heatmap) ---
plt.figure(figsize=(12, 10))
sns.heatmap(
    correlation_matrix,
    annot=True,      # Mostrar los valores numéricos en las celdas
    cmap='coolwarm', # Usar un mapa de color azul (negativo) a rojo (positivo)
    fmt='.2f',       # Formatear los números a dos decimales
    linewidths=.5
)
plt.title('Matriz de Correlación de Variables', fontsize=18)
plt.show()

print("\n--- Matriz de Correlación (Valores Numéricos) ---")
print(correlation_matrix)
```

**Interpretación del Análisis de Correlación**

La matriz de correlación es una tabla que muestra el coeficiente de correlación entre todas las combinaciones de variables. Este coeficiente es un valor entre -1 y 1 que indica la fuerza y dirección de una relación lineal:

**Cercano a 1 (Rojo)**: Correlación positiva fuerte. Cuando una variable aumenta, la otra también tiende a aumentar.

**Cercano a -1 (Azul)**: Correlación negativa fuerte. Cuando una variable aumenta, la otra tiende a disminuir.

**Cercano a 0 (Blanco/Gris)**: Correlación débil o nula. No hay una relación lineal clara entre las variables.

#5. Identificación de Posibles Sesgos y Patrones

**Sesgo de Género**: La fuerte dominancia masculina es un patrón real del TEA, pero también puede ocultar un sesgo de subdiagnóstico en mujeres, cuyas manifestaciones del espectro pueden ser diferentes y pasar desapercibidas.

**Sesgo Geográfico**: La alta concentración de casos en pocas comunas podría no reflejar la prevalencia real, sino una mayor capacidad de diagnóstico y reporte en esas áreas, o la migración de familias hacia zonas con mejores servicios. Las comunas con pocos casos podrían sufrir de un sesgo de sub-reporte.

**Sesgo de Edad**: La caída drástica de casos en la adolescencia tardía puede ser un sesgo de supervivencia en el sistema de datos, donde los estudiantes abandonan el sistema educativo formal y, por lo tanto, desaparecen de los registros.

**Patrón de Atención**: Existe un patrón claro de que el sistema educativo público es el principal responsable de la atención a la población con TEA en Cali.

#6. Hallazgos Clave e Implicaciones

**Hallazgo 1: Prevalencia Masculina**. La población estudiantil con TEA en Cali es predominantemente masculina, lo que concuerda con la literatura científica.

**Implicación**: Las estrategias pedagógicas y de apoyo deben estar bien informadas sobre las manifestaciones del TEA más comunes en hombres, sin descuidar la necesidad de capacitar a los docentes para identificar posibles casos en mujeres.

**Hallazgo 2: Concentración Geográfica**. La población no se distribuye uniformemente, con comunas específicas que concentran la mayoría de los casos registrados.

**Implicación**: La Secretaría de Educación puede focalizar la inversión y la creación de programas de apoyo (docentes de apoyo, terapias, etc.) en las comunas con mayor demanda. Es crucial investigar por qué otras comunas tienen tan pocos reportes.

**Hallazgo 3: Rol Central del Sector Público**. El sector Oficial atiende a la inmensa mayoría de estos estudiantes.

**Implicación**: Es imperativo fortalecer financieramente y técnicamente a las instituciones educativas oficiales. La formación docente, la adecuación de infraestructuras y la disponibilidad de material pedagógico son prioridades.

**Hallazgo 4: Concentración en Edades Escolares Básicas**. La mayoría de los estudiantes se encuentran en primaria y básica secundaria.

**Implicación**: Se deben desarrollar y fortalecer los programas de transición hacia la vida post-escolar (educación superior, formación técnica o vida laboral) para los jóvenes con TEA que están por salir del sistema.
